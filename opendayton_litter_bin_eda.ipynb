Description: Jupyter notebook that imports litter bin data and provides an overview of the data set.

```python
import duckdb
duckdb.sql("INSTALL spatial; LOAD spatial;")
```

```python
import pandas as pd
```

```python
import requests

url = 'https://services2.arcgis.com/3dDB2Kk6kuA2gIGw/arcgis/rest/services/Litter_container/FeatureServer/0/query?where=1%3D1&outFields=*&f=geojson&resultRecordCount=500' 
r = requests.get(url)

with open('litter.geojson', 'w') as f:
    f.write(r.text)
```

```python
df = duckdb.sql("""
SELECT * FROM ST_Read('litter.geojson')
""").df()
```

```python
for col in ['Condition', 'Rust', 'Door', 'Topper', 'Needs_Painting', 'Completed']:
       print(f"{col}: {df[col].value_counts(dropna=False).to_dict()}")
```

```python
print('Unique addresses:', df['Address'].nunique())
```

```python
for col in ['Rust', 'Door', 'Topper', 'Needs_Painting']:
       print(f"\n{col} by Condition:")
       print(pd.crosstab(df['Condition'], df[col], margins=True))
```

```python
pd.crosstab(df['Condition'], [df['Rust'], df['Door'], df['Topper'], df['Needs_Painting']])
```

```python
for col in ['Rust', 'Door', 'Topper', 'Needs_Painting']:
       print(f"\n{col} by Completed:")
       print(pd.crosstab(df['Completed'], df[col], margins=True))
```

```python
df['Completed'].value_counts()
```

```python
for col in ['Rust', 'Door', 'Topper', 'Needs_Painting']:
       print(f"{df.pivot_table(index='Completed', columns=col, aggfunc='size', fill_value=0, dropna=False)}\n")
```

```python
df.groupby('Completed')['GlobalID'].count()
```
